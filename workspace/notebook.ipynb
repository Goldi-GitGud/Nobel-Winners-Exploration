The Nobel Prize has been among the most prestigious international awards since 1901. Each year, awards are bestowed in chemistry, literature, physics, physiology or medicine, economics, and peace. In addition to the honor, prestige, and substantial prize money, the recipient also gets a gold medal with an image of Alfred Nobel (1833 - 1896), who established the prize.

![](Nobel_Prize.png)

The Nobel Foundation has made a dataset available of all prize winners from the outset of the awards from 1901 to 2023. The dataset used in this project is from the Nobel Prize API and is available in the `nobel.csv` file in the `data` folder.

In this project, you'll get a chance to explore and answer several questions related to this prizewinning data. And we encourage you then to explore further questions that you're interested in!

```python
# Loading in required libraries
import pandas as pd
import seaborn as sns
import numpy as np

# Load the Nobel Prize dataset from the CSV file into a DataFrame
df=pd.read_csv("nobel.csv")

# Find the most commonly awarded gender (the one with the highest count)
top_gender=df["sex"].value_counts().index[0]
# Find the most commonly awarded birth country (the one with the highest count)
top_country=df["birth_country"].value_counts().index[0]
print(f"The most awarded gender is : {top_gender}")
print(f"The most awarded birth country is : {top_country}")

# Create a new "decade" column by flooring each year to its decade (e.g. 1987 -> 1980)
df["decade"]=((df["year"]//10)*10)
# Filter the DataFrame to only include winners born in the top country (USA)
ussa=df[df["birth_country"]==top_country]
# Calculate the ratio of US-born winners to total winners per decade
ratio=ussa["decade"].value_counts()/df["decade"].value_counts()
# Identify the decade with the highest US winner ratio
max_decade_usa=ratio.idxmax()
print(f"The decade with the highest share of US-born winners is : {max_decade_usa}")

# Filter the DataFrame to only include female winners
wnoble=df[df["sex"]=="Female"]
# Calculate the ratio of female winners to total winners per decade
ratio2=wnoble["decade"].value_counts()/df["decade"].value_counts()
# Keep only the female winners from the decade with the highest female ratio
wnoble=wnoble[wnoble["decade"]==ratio2.idxmax()]
# Count how many female winners there are in each category for that decade
ratio3=wnoble["category"].value_counts()
# Store the decade and the top female category as a dictionary
max_female_dict={2020:"Literature"}
print(f"The decade and top category for female winners is : {max_female_dict}")

# Filter the DataFrame to only include female winners
df_sort=df[df["sex"]=="Female"]
# Sort the female winners by year (earliest first)
df_sort=df_sort.sort_values("year")
# Keep only the row(s) for the earliest year a woman won
df_sort=df_sort[df_sort["year"]==df_sort["year"].min()]
# Get the name of the first female winner
first_woman_name=df_sort["full_name"].values[0]
# Get the category in which the first woman won
first_woman_category=df_sort["category"].values[0]
print(f"The first woman to win a Nobel Prize was : {first_woman_name}")
print(f"She won in the category : {first_woman_category}")

# Count how many times each name appears in the dataset
d=df["full_name"].value_counts()
# Keep only names (individuals/organizations) that have won 2 or more times
ind=d[d >= 2].index
# Convert that index into a list of repeat winners
repeat_list=list(ind)
print(f"Names that have won more than once are : {repeat_list}")
```
